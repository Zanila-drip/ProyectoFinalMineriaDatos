```python
import requests
```

```python
CAMERA_PORT = 8001  # Cambia según la cámara: 8001, 8002, 8003, 8004
response = requests.get(f"http://localhost:{CAMERA_PORT}/api/v1/videos")
print(response.json())
```

```python
def upload_video(video_path: str, camera_id: str):
    """
    Sube un video a la API

    Args:
        video_path: Ruta al archivo de video
        camera_id: ID de la cámara (cam1, cam2, cam3, cam4)
    """
    url = f"http://143.198.59.209:8000/api/v1/videos/upload/{camera_id}"

    with open(video_path, 'rb') as video_file:
        files = {'video': (video_path.split('/')[-1], video_file, 'video/mp4')}
        response = requests.post(url, files=files)

    if response.status_code == 200:
        print("Video subido exitosamente")
        print(response.json())
    else:
        print(f"Error al subir el video: {response.status_code}")
        print(response.text)
```
